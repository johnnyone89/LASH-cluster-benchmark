# 04 — Final Robustness, Architecture, and Storage Analyses

Run this notebook after Notebook 01. It executes the audited robustness engine already included in the repository, but uses repository-relative paths and writes the additional results under `outputs/final_robustness/`. The workflow covers 10-seed focal inference, 24/72/168/336-h dependence scales, reduced architecture, retained-component tests, external-scope audit, and constrained storage dispatch.


```python
from pathlib import Path
import json, nbformat

ROOT = Path.cwd().resolve()
if ROOT.name.lower() == 'notebooks':
    ROOT = ROOT.parent
OUTPUT_DIR = ROOT / 'outputs'
engine_nb = ROOT / 'notebooks' / '04_SecondRound_Confirmatory.ipynb'
if not engine_nb.exists():
    raise FileNotFoundError(engine_nb)

# Compatibility marker for the audited robustness engine. The final benchmark
# itself is generated by Notebook 01 with the same dataset-local selection boundary.
final_contract = OUTPUT_DIR / 'final_run_contract.json'
if not final_contract.exists():
    raise FileNotFoundError('Run 01_Final_10Seed_Benchmark.ipynb first.')
contract = json.loads(final_contract.read_text(encoding='utf-8'))
compat = dict(contract)
compat['protocol'] = 'reviewer_20260823_per_dataset_hpo_v1'
compat['test_used_for_selection'] = False
(OUTPUT_DIR / 'run_contract_20260823_per_dataset_hpo.json').write_text(json.dumps(compat, indent=2), encoding='utf-8')
print('Robustness engine:', engine_nb)
```

```python
nb = nbformat.read(engine_nb, as_version=4)
code_cells = [c.source for c in nb.cells if c.cell_type == 'code']
print('Executing', len(code_cells), 'audited analysis cells...')
for i, source in enumerate(code_cells, 1):
    # Keep computation unchanged; only redirect the auxiliary output folder.
    source = source.replace('second_round_confirmatory', 'final_robustness')
    print(f'[{i:02d}/{len(code_cells):02d}]')
    exec(compile(source, f'<robustness-cell-{i}>', 'exec'), globals(), globals())
print('Final robustness workflow complete.')
```
